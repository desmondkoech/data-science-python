# Population visualization
This notebook demonstrates a simple bar chart generated from sample data.


```
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame({'Country': ['A','B','C'], 'Population': [10,20,15]})
ax = df.plot(kind='bar', x='Country', y='Population', legend=False)
fig = ax.get_figure()
fig.savefig('../outputs/population.png')
plt.close(fig)
```
